```python
import math

# Reprezentacja planszy: lista 9 elementów (indeksy 0-8)
plansza = [' ' for _ in range(9)]

def drukuj_plansze(p):
    print(f"\n {p[0]} | {p[1]} | {p[2]} ")
    print("---+---+---")
    print(f" {p[3]} | {p[4]} | {p[5]} ")
    print("---+---+---")
    print(f" {p[6]} | {p[7]} | {p[8]} \n")

def sprawdz_wygrana(p, gracz):
    # Możliwe kombinacje wygrywające
    win_states = [
        (0, 1, 2), (3, 4, 5), (6, 7, 8), # Poziomo
        (0, 3, 6), (1, 4, 7), (2, 5, 8), # Pionowo
        (0, 4, 8), (2, 4, 6)             # Na ukos
    ]
    for a, b, c in win_states:
        if p[a] == p[b] == p[c] == gracz:
            return True
    return False

def czy_pelna(p):
    return ' ' not in p

# Główny algorytm Minimax z Alpha-Beta Pruning
def minimax_ab(p, depth, is_maximizing, alpha, beta):
    # Warunki końcowe rekurencji
    if sprawdz_wygrana(p, 'O'): return 1   # Bot wygrywa
    if sprawdz_wygrana(p, 'X'): return -1  # Człowiek wygrywa
    if czy_pelna(p): return 0              # Remis

    if is_maximizing: # Ruch Bota
        best_score = -math.inf
        for i in range(9):
            if p[i] == ' ':
                p[i] = 'O'
                score = minimax_ab(p, depth + 1, False, alpha, beta)
                p[i] = ' ' # Cofnij ruch (backtracking)
                best_score = max(score, best_score)
                alpha = max(alpha, best_score)
                if beta <= alpha: # Cięcie (Pruning)
                    break
        return best_score
    else: # Ruch Człowieka (symulowany)
        best_score = math.inf
        for i in range(9):
            if p[i] == ' ':
                p[i] = 'X'
                score = minimax_ab(p, depth + 1, True, alpha, beta)
                p[i] = ' ' # Cofnij ruch
                best_score = min(score, best_score)
                beta = min(beta, best_score)
                if beta <= alpha: # Cięcie (Pruning)
                    break
        return best_score

def ruch_bota():
    best_score = -math.inf
    move = -1
    # Bot szuka najlepszego ruchu
    for i in range(9):
        if plansza[i] == ' ':
            plansza[i] = 'O'
            score = minimax_ab(plansza, 0, False, -math.inf, math.inf)
            plansza[i] = ' '
            if score > best_score:
                best_score = score
                move = i
    plansza[move] = 'O'
    print(f"Bot wybrał pole: {move}")

def gra():
    print("Witaj w grze Kółko i Krzyżyk z algorytmem Alpha-Beta!")
    print("Pola są numerowane od 0 do 8 (rzędami).")
    
    start = input("Kto zaczyna? (wpisz 'ja' lub 'bot'): ").strip().lower()
    tura_czlowieka = True if start == 'ja' else False

    while True:
        drukuj_plansze(plansza)

        if sprawdz_wygrana(plansza, 'O'):
            print("Bot wygrał! (To było do przewidzenia...)")
            break
        if sprawdz_wygrana(plansza, 'X'):
            print("Wygrałeś! (To niemożliwe przy poprawnym algorytmie)")
            break
        if czy_pelna(plansza):
            print("Remis!")
            break

        if tura_czlowieka:
            try:
                ruch = int(input("Twój ruch (1-9): "))-1
                if plansza[ruch] == ' ':
                    plansza[ruch] = 'X'
                    tura_czlowieka = False
                else:
                    print("To pole jest zajęte!")
            except (ValueError, IndexError):
                print("Nieprawidłowy wpis. Podaj liczbę 0-8.")
        else:
            print("Bot myśli...")
            ruch_bota()
            tura_czlowieka = True

if __name__ == "__main__":
    gra()
```

```python
import math

# Początkowy stan planszy
plansza = [' ' for i in range(9)]

def drukuj_plansze(stan):
    print(f"\n {stan[0]} | {stan[1]} | {stan[2]} ")
    print("---+---+---")
    print(f" {stan[3]} | {stan[4]} | {stan[5]} ")
    print("---+---+---")
    print(f" {stan[6]} | {stan[7]} | {stan[8]} \n")

# Pomocnicza funkcja do sprawdzania zwycięzcy (używana w koniec i wartosc)
def sprawdz_zwyciezce(stan, gracz):
    # Poprawione indeksy (0-8)
    poz_wygrywajace = [
        (0,1,2), (3,4,5), (6,7,8), # wiersze
        (0,3,6), (1,4,7), (2,5,8), # kolumny
        (0,4,8), (2,4,6)           # diagonale
    ]
    for a, b, c in poz_wygrywajace:
        if stan[a] == stan[b] == stan[c] == gracz:
            return True
    return False

# 1. Funkcja: Czy gra się skończyła?
def koniec(stan):
    # Koniec jeśli wygrał X, wygrał O, lub nie ma pustych pól (remis)
    if sprawdz_zwyciezce(stan, 'X'):
        return True
    if sprawdz_zwyciezce(stan, 'O'):
        return True
    if ' ' not in stan:
        return True
    return False

# 2. Funkcja: Jaka jest wartość stanu końcowego?
def wartosc(stan):
    if sprawdz_zwyciezce(stan, 'O'):
        return 1  # Wygrał MAX (Bot)
    elif sprawdz_zwyciezce(stan, 'X'):
        return -1 # Wygrał MIN (Człowiek)
    else:
        return 0  # Remis

# 3. Funkcja: Czyja jest teraz tura? (MAX czy MIN)
def gracz(stan):
    liczba_x = stan.count('X')
    liczba_o = stan.count('O')
    # Zakładamy, że X zaczyna. Jeśli jest tyle samo X i O, to ruch ma X.
    # Jeśli X jest więcej, to ruch ma O.
    if liczba_x == liczba_o:
        return 'MIN' # Ruch Gracza X
    else:
        return 'MAX' # Ruch Gracza O (Bot)

# 4. Funkcja: Zwraca listę możliwych ruchów (indeksów)
def akcje(stan):
    mozliwe_ruchy = []
    for i in range(9):
        if stan[i] == ' ':
            mozliwe_ruchy.append(i)
    return mozliwe_ruchy

# 5. Funkcja: Zwraca NOWY stan po wykonaniu akcji
def wynik(stan, akcja):
    # Ważne: Tworzymy kopię planszy, żeby nie psuć obecnego stanu w rekurencji!
    nowy_stan = list(stan)
    
    # Sprawdzamy kto ma ruch na podstawie starego stanu
    kto_rusza = gracz(stan) 
    symbol = 'O' if kto_rusza == 'MAX' else 'X'
    
    nowy_stan[akcja] = symbol
    return nowy_stan

# --- ALGORYTM MINIMAX ---

def minimax(stan):
    # Warunek stopu rekurencji
    if koniec(stan):
        return wartosc(stan)

    kto = gracz(stan)

    if kto == 'MAX': # Tura Bota ('O') - szukamy największej wartości
        v = -math.inf
        for a in akcje(stan):
            # Rekurencyjnie sprawdzamy wynik każdego ruchu
            v_kandydata = minimax(wynik(stan, a))
            v = max(v, v_kandydata)
        return v
    
    else: # Tura Człowieka ('X') - symulujemy, że zagra najlepiej dla siebie (najmniej)
        v = math.inf
        for a in akcje(stan):
            v_kandydata = minimax(wynik(stan, a))
            v = min(v, v_kandydata)
        return v

# --- FUNKCJA STERUJĄCA (BOT WYBIERA RUCH) ---

def ruch_bota_minimax(obecny_stan):
    najlepszy_wynik = -math.inf
    najlepsza_akcja = None
    
    # Sprawdzamy każdy dostępny ruch
    for a in akcje(obecny_stan):
        # Symulujemy ruch
        stan_po_ruchu = wynik(obecny_stan, a)
        # Obliczamy wartość tego ruchu za pomocą minimax
        wartosc_ruchu = minimax(stan_po_ruchu)
        
        # Jeśli ten ruch jest lepszy niż dotychczasowy najlepszy, zapamiętujemy go
        if wartosc_ruchu > najlepszy_wynik:
            najlepszy_wynik = wartosc_ruchu
            najlepsza_akcja = a
            
    return najlepsza_akcja

# --- PĘTLA GRY ---

if __name__ == "__main__":
    print("Gra: Ty jesteś 'X' (zaczynasz), Bot to 'O'.")
    aktualna_plansza = plansza
    
    while not koniec(aktualna_plansza):
        drukuj_plansze(aktualna_plansza)
        
        turn = gracz(aktualna_plansza)
        
        if turn == 'MIN': # Twój ruch
            try:
                wybor = int(input("Twój ruch (0-8): "))
                if aktualna_plansza[wybor] == ' ':
                    aktualna_plansza = wynik(aktualna_plansza, wybor)
                else:
                    print("To pole jest zajęte!")
            except (ValueError, IndexError):
                print("Nieprawidłowy numer pola.")
        
        else: # Ruch bota
            print("Bot myśli...")
            ruch = ruch_bota_minimax(aktualna_plansza)
            aktualna_plansza = wynik(aktualna_plansza, ruch)

    drukuj_plansze(aktualna_plansza)
    val = wartosc(aktualna_plansza)
    if val == 1: print("Przegrałeś! Bot wygrał.")
    elif val == -1: print("Wygrałeś! (Niemożliwe przy poprawnym Minimaxie)")
    else: print("Remis!")
```
